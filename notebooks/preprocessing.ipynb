### Normalization, Scaling & Encoding for UPI Transactions

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.data_preprocessing import UPITransactionPreprocessor, PreprocessingConfig
from src.utils import MERGED_DATA_DIR

parquet_path = PROJECT_ROOT / "data" / "merged" / "mapped_common_schema.parquet"
df = pd.read_parquet(parquet_path).head(10000)
print(f"Sample loaded: {df.shape}")
```

## Preprocessor Initialization & Fitting

```python
config = PreprocessingConfig(scaler="robust", encoder="onehot")
preprocessor = UPITransactionPreprocessor(config)

X_trans, y = preprocessor.fit_transform(df)
feature_names = preprocessor.get_feature_names()

print("Transformed Feature Matrix Shape:", X_trans.shape)
print("Extracted Feature Count:", len(feature_names))
print("Sample Features:", feature_names[:10])
```
